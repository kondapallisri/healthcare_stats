```python
import numpy as np
import pandas as pd
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Section A
## 1: Central Tendency of Patient Admissions
patient_admissions = [32, 28, 35, 30, 29, 27, 31, 34, 33, 30]
mean_admissions = np.mean(patient_admissions)
median_admissions = np.median(patient_admissions)
mode_admissions = stats.mode(patient_admissions, keepdims=True).mode[0]
print("Mean Admissions:", mean_admissions)
print("Median Admissions:", median_admissions)
print("Mode Admissions:", mode_admissions)
```

```python
# Increase admissions by 10%
increased_admissions = [int(x * 1.1) for x in patient_admissions]
mean_increased = np.mean(increased_admissions)
median_increased = np.median(increased_admissions)
mode_increased = stats.mode(increased_admissions, keepdims=True).mode[0]
print("Mean Increased Admissions:", mean_increased)
print("Median Increased Admissions:", median_increased)
print("Mode Increased Admissions:", mode_increased)
```

```python
## 2: Variability in Recovery Duration
recovery_durations = [5, 7, 6, 8, 9, 5, 6, 7, 8, 6]
range_recovery = np.ptp(recovery_durations)
variance_recovery = np.var(recovery_durations, ddof=1)
std_dev_recovery = np.std(recovery_durations, ddof=1)
print("Range of Recovery Duration:", range_recovery)
print("Variance of Recovery Duration:", variance_recovery)
print("Standard Deviation of Recovery Duration:", std_dev_recovery)
```

```python
# Adding two new recovery durations
new_recovery_durations = recovery_durations + [4, 10]
new_std_dev = np.std(new_recovery_durations, ddof=1)
print("New Standard Deviation (After Adding Two Patients):", new_std_dev)
```

```python
## 3: Skewness and Kurtosis of Satisfaction Scores
satisfaction_scores = [8, 9, 7, 8, 10, 7, 9, 6, 10, 8, 7, 9]
skewness = stats.skew(satisfaction_scores)
kurtosis = stats.kurtosis(satisfaction_scores)
print("Skewness of Satisfaction Scores:", skewness)
print("Kurtosis of Satisfaction Scores:", kurtosis)
```

```python
## 4: Correlation Between Nurses and Recovery Time
nurses = [10, 12, 15, 18, 20, 22]
recovery_time = [8, 7, 6, 5, 4, 3]
correlation = np.corrcoef(nurses, recovery_time)[0, 1]
print("Correlation between Nurses and Recovery Time:", correlation)
```

```python
# Section B
## 5: Hypothesis Testing for Wait Time
wait_times = [32, 29, 31, 34, 33, 27, 30, 28, 35, 26]
population_mean = 30
alpha = 0.05
t_stat, p_value = stats.ttest_1samp(wait_times, population_mean)
print("T-statistic for Wait Time Hypothesis Test:", t_stat)
print("P-value for Wait Time Hypothesis Test:", p_value)
```

```python
## 6: Chi-Square Test for Cleanliness vs Satisfaction
data = np.array([[90, 10], [60, 40], [30, 70]])
chi2, p_cleanliness, _, _ = stats.chi2_contingency(data)
print("Chi-Square Test Statistic for Cleanliness vs Satisfaction:", chi2)
print("P-value for Cleanliness vs Satisfaction:", p_cleanliness)
```

```python
## 7: ANOVA for Treatment Methods
treatment_A = [5, 6, 7, 5, 6]
treatment_B = [8, 9, 7, 8, 10]
treatment_C = [4, 5, 6, 5, 4]
f_stat, p_treatment = stats.f_oneway(treatment_A, treatment_B, treatment_C)
print("ANOVA Test Statistic for Treatment Methods:", f_stat)
print("P-value for Treatment Methods:", p_treatment)
```

```python
## 8: Normality Test for Administration Times
admin_times = [12, 15, 14, 16, 18, 13, 14, 17, 15, 19, 16, 14]
normality_test = stats.shapiro(admin_times)
print("Shapiro-Wilk Test for Normality (Administration Times):", normality_test)
```

```python
## 9: Poisson Distribution for Emergency Arrivals
lambda_value = 5
prob_3_arrivals = stats.poisson.pmf(3, lambda_value)
print("Probability of Exactly 3 Emergency Arrivals:", prob_3_arrivals)
```

```python
## 10: Expected Value for Surgeries
surgeries = np.array([0, 1, 2, 3, 4, 5])
frequencies = np.array([5, 12, 18, 22, 15, 8])
p_values = frequencies / sum(frequencies)
expected_surgeries = np.sum(surgeries * p_values)
print("Expected Number of Surgeries Performed per Day:", expected_surgeries)
```
